# Chapter 2 — Burgers (sawtooth): TD-PINs vs Vanilla PINs — **KAN + adaptive grid refinement**

**Case:** 1-D viscous Burgers, sawtooth analytic solution (Hopf–Cole) on $[0, 2\pi]$,
$\;u_t + u\,u_x = \nu\,u_{xx}$.

Same experiment as `burgers_tdpinns_vs_vanilla_sawtooth_kan.ipynb`, with one addition: an **optional
adaptive grid-refinement mode** on the KAN backbone.

## The refinement mode

KANs represent each activation as a B-spline on a grid of `grid_size` intervals. A coarse grid trains
fast but caps the resolvable detail (e.g. the shock). `pykan`'s grid extension (`MultKAN.refine`) makes
the grid finer while preserving the learned function. We expose this as a **training-time policy carried
in the architecture params** (`PARAMS_TD` / `PARAMS_VAN`):

| kwarg | meaning |
|---|---|
| `refine_every` | check for refinement every $N$ epochs (`None` ⇒ feature **off**) |
| `refine_factor` | grid-size multiplier per refinement (default `1.1` ⇒ **+10 %**) |
| `refine_tol` | minimum **relative loss improvement** expected over a window; refine **only** when the actual improvement is below it |
| `refine_max_grid` | hard cap on grid intervals (`None` ⇒ uncapped) |

**Efficiency:** refinement is *adaptive*, not periodic — the grid grows only when the loss is **not
reducing fast enough** (window improvement `< refine_tol`). While the net is still making good progress
no capacity is added, so we don't pay for splines we don't need.

Reporting is unchanged: error norms ($L_1,L_2,L_\infty$), conservation (mass/energy), and per-band
spectral error vs the analytical sawtooth.

```python
import sys, os, time, shutil

ROOT = '/home/df.gomezb/Tesis'
if ROOT not in sys.path:
    sys.path.insert(0, ROOT)

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

from pinns_framework import (
    PINArchitecture, PINSpatialDimension, PINLossFunction, PINTrainer,
    compute_metrics, collocation_uniform,
)

torch.manual_seed(0)
np.random.seed(0)

plt.rcParams.update({
    'font.family'     : 'serif',
    'font.serif'      : ['DejaVu Serif'],
    'mathtext.fontset': 'cm',
    'font.size'       : 12,
    'axes.titlesize'  : 13,
    'axes.labelsize'  : 12,
    'legend.fontsize' : 9,
})

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('torch:', torch.__version__, '| device:', DEVICE)

METHODS = {'TD-PINs': ('o', 'tab:blue'), 'Vanilla': ('s', 'tab:orange')}

FIGDIR = 'figs'
os.makedirs(FIGDIR, exist_ok=True)

def clear_kan_cache():
    """pykan writes a ./model checkpoint dir on every KAN construction / refinement."""
    shutil.rmtree('./model', ignore_errors=True)

def refine_report(trainer, label):
    """Pretty-print how many grid refinements each training call performed."""
    n = sum(h.get('refinements', 0) for h in trainer.step_history)
    per = [h.get('refinements', 0) for h in trainer.step_history]
    print(f'{label}: {n} grid refinement(s) total | per call: {per}')
```

## 1. Problem setup — sawtooth analytic solution

```python
NU     = 0.07
X0, XF = 0.0, 2 * np.pi

def toothsaw(t, x):
    """Hopf-Cole analytic solution for viscous Burgers on [0, 2pi]."""
    x  = np.asarray(x, dtype=float)
    nu = NU
    c  = 4 * nu * (t + 1)
    a  = (-4*t + x)**2           / c
    b  = (-4*t + x - 2*np.pi)**2 / c
    num = ((-8*t + 2*x)            * np.exp(-a)
         + (-8*t + 2*x - 4*np.pi) * np.exp(-b)) / c
    den = np.exp(-b) + np.exp(-a)
    return 2 * nu * num / den + 4

x_plot = np.linspace(X0, XF, 500)
fig, ax = plt.subplots(figsize=(8, 3.5))
cmap = plt.cm.plasma
for i, t in enumerate([0.0, 0.01, 0.02, 0.025]):
    ax.plot(x_plot, toothsaw(t, x_plot), color=cmap(i / 3), lw=1.8, label=f't = {t:.3f}')
ax.set_xlabel('x'); ax.set_ylabel('u')
ax.set_title('Sawtooth analytic solution  (nu = 0.07)')
ax.legend(loc='upper right'); ax.grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

## 2. Experiment configuration

KAN backbone with the **adaptive grid-refinement policy** added to both param dicts. Start from a coarse
grid (`grid_size=5`) so refinement has room to act; the grid grows by `refine_factor` (+10 %) each time
the loss stalls, up to `REFINE_MAXG`. Set `REFINE_EVERY = None` to recover the plain KAN run.

```python
# ── adaptive grid-refinement policy (shared by both methods) ──────────────
REFINE_EVERY  = 1500     # check every N epochs (None disables refinement)
REFINE_FACTOR = 1.10     # +10% grid intervals per refinement (default)
REFINE_TOL    = 0.20     # refine only if window improvement < 20% (loss stalling)
REFINE_MAXG   = 20       # max grid intervals allowed (cap)

_REFINE = {'refine_every': REFINE_EVERY, 'refine_factor': REFINE_FACTOR,
           'refine_tol': REFINE_TOL, 'refine_max_grid': REFINE_MAXG}

# KAN backbones — start coarse (grid_size=5) so refinement has headroom.
PARAMS_TD  = {'layers': (1, 16, 16, 1), 'grid_size': 5, 'spline_order': 3,
              'grid_range': (X0, XF),  'grid_eps': 0.05, **_REFINE}   # u(x)
PARAMS_VAN = {'layers': (2, 16, 16, 1), 'grid_size': 5, 'spline_order': 3,
              'grid_range': (0.0, XF), 'grid_eps': 0.05, **_REFINE}   # u(t, x)

DT     = 0.005
T_END  = 0.025
NT     = round(T_END / DT)
times  = [k * DT for k in range(NT + 1)]

# optimisation (shared schedule)
LR, MAX_LR, MIN_LR = 1e-2, 1e-2, 1e-6
_norm      = (15**2 + 5**2 + 1**2) ** 0.5
IC_WEIGHT  = 15.0 / _norm
BC_WEIGHT  = 5.0  / _norm
PDE_WEIGHT = 1.0  / _norm
N_COLLOC   = 500
N_EVAL     = 1000

TD_EPOCHS,  TD_TOL  = 8000,  1e-5     # CN per-step
VAN_EPOCHS, VAN_TOL = 20000, 1e-5     # space-time fit

print(f'{NT} steps, dt={DT}, T={T_END}')
if REFINE_EVERY:
    print(f'refinement: every {REFINE_EVERY} ep | x{REFINE_FACTOR} grid | '
          f'trigger improvement < {REFINE_TOL:.0%} | cap {REFINE_MAXG}')
else:
    print('refinement: OFF')
```

## 3. IC / BC, residuals, exact evaluator, time-slice wrapper

- **TD residual** (Crank–Nicolson): $\frac{u-u^{t-1}}{\Delta t} + \tfrac12[S(u)+S(u^{t-1})]$, $S(w)=w w_x-\nu w_{xx}$.
- **Vanilla residual**: $u_t + u u_x - \nu u_{xx}$.
- **`TimeSlice`** wraps the Vanilla $u(t,x)$ KAN into a spatial-only $u(x)$ at a fixed $t$.

```python
x_ic_np = np.linspace(X0, XF, 201)
u_ic_np = toothsaw(0, x_ic_np)
x_ic = torch.tensor(x_ic_np, dtype=torch.float32).unsqueeze(1)
u_ic = torch.tensor(u_ic_np, dtype=torch.float32).unsqueeze(1)

u_bc_val = float(toothsaw(0, X0))
x_bc = torch.tensor([[X0], [XF]], dtype=torch.float32)
u_bc = torch.full((2, 1), u_bc_val, dtype=torch.float32)

x_dim = PINSpatialDimension('x', domain=(X0, XF), n_colloc=N_COLLOC)

def td_residual(u, u_prev, x, dt, nu):
    """TD-PINs Crank-Nicolson residual."""
    def S(w):
        w_x  = torch.autograd.grad(w.sum(),   x, create_graph=True)[0]
        w_xx = torch.autograd.grad(w_x.sum(), x, create_graph=True)[0]
        return w * w_x - nu * w_xx
    return (u - u_prev.detach()) / dt + 0.5 * (S(u) + S(u_prev).detach())

def vanilla_residual(u, t, x, nu):
    """Vanilla strong-form Burgers residual u_t + u u_x - nu u_xx."""
    u_t  = torch.autograd.grad(u.sum(),   t, create_graph=True)[0]
    u_x  = torch.autograd.grad(u.sum(),   x, create_graph=True)[0]
    u_xx = torch.autograd.grad(u_x.sum(), x, create_graph=True)[0]
    return u_t + u * u_x - nu * u_xx

loss_td  = PINLossFunction([x_dim], td_residual)
loss_van = PINLossFunction([x_dim], vanilla_residual)

def exact_fn(t, x_pts):
    u = toothsaw(t, x_pts.detach().cpu().numpy().ravel())
    return torch.tensor(u, dtype=torch.float32, device=x_pts.device).unsqueeze(1)

class TimeSlice(nn.Module):
    """Expose a space-time net u(t,x) as a spatial-only net u(x) at fixed t."""
    def __init__(self, net, t):
        super().__init__(); self.net = net; self.t = float(t)
    def forward(self, x):
        tcol = torch.full((x.shape[0], 1), self.t, device=x.device, dtype=x.dtype)
        return self.net(torch.cat([tcol, x], dim=1))
```

## 4. Train TD-PINs (Crank–Nicolson) — KAN per step, with refinement

```python
clear_kan_cache()
torch.manual_seed(0); np.random.seed(0)
net_td = PINArchitecture('KAN', PARAMS_TD)
trainer_td = PINTrainer(
    net=net_td, scheme='TD-PINNS', loss_fn=loss_td, spatial_dims=[x_dim],
    epochs=TD_EPOCHS, dt=DT, lr=LR, max_lr=MAX_LR, min_lr=MIN_LR, tolerance=TD_TOL,
    ic_weight=IC_WEIGHT, bc_weight=BC_WEIGHT, pde_weight=PDE_WEIGHT,
    collocation_fn=collocation_uniform, device=DEVICE, verbose=True, log_every=2000,
)
t0 = time.perf_counter()
nets_td = trainer_td.fit(x_ic=x_ic, u_ic=u_ic, x_bc=x_bc, u_bc=u_bc, T=T_END, nu=NU)
wall_td = time.perf_counter() - t0
epochs_td = sum(h['epochs'] for h in trainer_td.step_history[1:])
print(f'TD-PINs (KAN) done | epochs(PDE steps)={epochs_td} | wall {wall_td:.1f}s')
refine_report(trainer_td, 'TD-PINs')
print('final grid per net:', [int(n.net.grid) for n in nets_td])
```

## 5. Train Vanilla PINs (single space–time KAN, with refinement)

```python
clear_kan_cache()
torch.manual_seed(0); np.random.seed(0)
net_van = PINArchitecture('KAN', PARAMS_VAN)
trainer_van = PINTrainer(
    net=net_van, scheme='VANILLA', loss_fn=loss_van, spatial_dims=[x_dim],
    epochs=VAN_EPOCHS, lr=LR, max_lr=MAX_LR, min_lr=MIN_LR, tolerance=VAN_TOL,
    ic_weight=IC_WEIGHT, bc_weight=BC_WEIGHT, pde_weight=PDE_WEIGHT,
    collocation_fn=collocation_uniform, device=DEVICE, verbose=True, log_every=2000,
)
t0 = time.perf_counter()
vanilla_net = trainer_van.fit(x_ic=x_ic, u_ic=u_ic, x_bc=x_bc, u_bc=u_bc, T=T_END, nu=NU)[0]
wall_van = time.perf_counter() - t0
epochs_van = trainer_van.step_history[-1]['epochs']
print(f'Vanilla (KAN) done | epochs={epochs_van} | wall {wall_van:.1f}s')
refine_report(trainer_van, 'Vanilla')
print('final grid:', int(vanilla_net.net.grid))
```

## 6. Evaluate both methods — all error norms ($L_1$, $L_2$, $L_\infty$)

```python
vanilla_slices = [TimeSlice(vanilla_net, t).to(DEVICE) for t in times]

def add_L1(df, slice_nets):
    xg = torch.linspace(X0, XF, N_EVAL).unsqueeze(1).to(DEVICE)
    l1 = []
    for t, net in zip(times, slice_nets):
        net.eval()
        with torch.no_grad():
            up = net(xg)
            if up.dim() == 1:
                up = up.unsqueeze(1)
        l1.append(float((up - exact_fn(t, xg)).abs().mean()))
    df['L1'] = l1
    return df

results = {
    'TD-PINs': {'slices': nets_td,        'epochs': epochs_td,  'wall': wall_td},
    'Vanilla': {'slices': vanilla_slices, 'epochs': epochs_van, 'wall': wall_van},
}
for m, r in results.items():
    df = compute_metrics(r['slices'], times, [x_dim], n_eval=N_EVAL,
                         exact_fn=exact_fn, device=DEVICE)
    r['df'] = add_L1(df, r['slices'])

rows = []
for m, r in results.items():
    d = r['df']
    rows.append({'method': m, 'epochs': r['epochs'],
                 'mean_L1': d['L1'].mean(), 'mean_L2': d['L2'].mean(), 'mean_Linf': d['Linf'].mean(),
                 'final_L1': d['L1'].iloc[-1], 'final_L2': d['L2'].iloc[-1], 'final_Linf': d['Linf'].iloc[-1],
                 'wall_s': r['wall']})
summary = pd.DataFrame(rows).set_index('method')
pd.set_option('display.float_format', '{:.4e}'.format)
print('=== TD-PINs vs Vanilla (KAN + refinement) — error summary (vs analytical) ===')
print(summary.to_string())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.2))
for ax, (label, col) in zip(axes, [(r'$L_1$', 'L1'), (r'$L_2$', 'L2'), (r'$L_\infty$', 'Linf')]):
    for m, r in results.items():
        mk, c = METHODS[m]
        ax.semilogy(times, r['df'][col].values, '-', marker=mk, ms=6, lw=1.5, color=c, label=m)
    ax.set_xlabel('t'); ax.set_ylabel(f'{label} error vs exact')
    ax.set_title(f'{label} error per time level')
    ax.grid(True, which='both', alpha=0.3); ax.legend()
fig.suptitle('Burgers sawtooth (KAN + refinement) — TD-PINs vs Vanilla', fontsize=14)
plt.tight_layout()
fig.savefig(os.path.join(FIGDIR, 'burgers_kan_refine_methods_errors.pdf'), bbox_inches='tight')
print('saved ->', os.path.join(FIGDIR, 'burgers_kan_refine_methods_errors.pdf'))
plt.show()
```

## 7. Conservation — mass $\int u\,dx$ and energy $\int u^2\,dx$

Mass should stay flat ($\int u\,dx = 4\cdot 2\pi$); energy must decrease monotonically (viscous
dissipation $dE/dt = -2\nu\int u_x^2\,dx \le 0$).

```python
df_ref = results['TD-PINs']['df']

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

ax = axes[0]
ax.plot(times, df_ref['exact_mass'].values, 'k-', lw=2.4, label='analytical')
for m, r in results.items():
    mk, c = METHODS[m]
    ax.plot(times, r['df']['mass'].values, '-', marker=mk, ms=6, lw=1.5, color=c, label=m)
ax.set_xlabel('t'); ax.set_ylabel(r'mass  $\int u\,dx$')
ax.set_title('Mass conservation (flat = conserved)')
ax.grid(alpha=0.3); ax.legend()

ax = axes[1]
ax.plot(times, df_ref['exact_energy'].values, 'k-', lw=2.4, label='analytical')
for m, r in results.items():
    mk, c = METHODS[m]
    ax.plot(times, r['df']['energy'].values, '-', marker=mk, ms=6, lw=1.5, color=c, label=m)
ax.set_xlabel('t'); ax.set_ylabel(r'energy  $\int u^2\,dx$')
ax.set_title(r'Energy dissipation ($dE/dt \leq 0$, decreasing)')
ax.grid(alpha=0.3); ax.legend()

plt.tight_layout()
fig.savefig(os.path.join(FIGDIR, 'burgers_kan_refine_methods_conservation.pdf'), bbox_inches='tight')
print('saved ->', os.path.join(FIGDIR, 'burgers_kan_refine_methods_conservation.pdf'))
plt.show()

print('\nanalytical mass (should be constant):', np.round(df_ref['exact_mass'].values, 5))
print('analytical energy (should decrease) :', np.round(df_ref['exact_energy'].values, 4))
```

## 8. Spectral test — error per frequency band

FFT in $x$ at the final time, error grouped into DC / low / mid / high bands. The shock lives in the
**high** wavenumbers — the band that benefits most from a finer (refined) grid.

```python
N_SPEC = 512
x_spec = np.linspace(X0, XF, N_SPEC, endpoint=False)
xs     = torch.tensor(x_spec, dtype=torch.float32).unsqueeze(1).to(DEVICE)
kf     = NT
t_spec = times[kf]

u_ex = toothsaw(t_spec, x_spec)
U_ex = np.fft.rfft(u_ex) / N_SPEC
kk   = np.arange(len(U_ex))

spec = {}
for m, r in results.items():
    net = r['slices'][kf].eval()
    with torch.no_grad():
        u = net(xs).cpu().numpy().ravel()
    spec[m] = np.fft.rfft(u) / N_SPEC

BANDS = [(0, 0, 'DC (k=0)'), (1, 4, 'low (1-4)'), (5, 15, 'mid (5-15)'), (16, kk[-1], 'high (16+)')]
band_labels = [b[2] for b in BANDS]
band_err = {m: [np.abs(spec[m] - U_ex)[lo:hi + 1].sum() for lo, hi, _ in BANDS] for m in results}

fig, axes = plt.subplots(1, 3, figsize=(16, 4.4))

ax = axes[0]
ax.semilogy(kk[1:], np.abs(U_ex)[1:], 'k-', lw=2.0, label='analytical')
for m in results:
    mk, c = METHODS[m]
    ax.semilogy(kk[1:], np.abs(spec[m])[1:], '-', color=c, lw=1.3, label=m)
ax.set_xlabel('wavenumber k'); ax.set_ylabel(r'$|\hat u_k|$')
ax.set_title(f'Spectrum at t = {t_spec:.3f}'); ax.set_xlim(0, 40)
ax.grid(True, which='both', alpha=0.3); ax.legend()

ax = axes[1]
for m in results:
    mk, c = METHODS[m]
    ax.semilogy(kk[1:], np.abs(spec[m] - U_ex)[1:], '-', color=c, lw=1.3, label=m)
ax.set_xlabel('wavenumber k'); ax.set_ylabel(r'$|\hat u^{pred}_k - \hat u^{exact}_k|$')
ax.set_title('Per-mode error'); ax.set_xlim(0, 40)
ax.grid(True, which='both', alpha=0.3); ax.legend()

ax = axes[2]
xb = np.arange(len(BANDS)); w = 0.38
for i, m in enumerate(results):
    mk, c = METHODS[m]
    ax.bar(xb + (i - 0.5) * w, band_err[m], width=w, color=c, label=m)
ax.set_yscale('log')
ax.set_xticks(xb); ax.set_xticklabels(band_labels, rotation=15)
ax.set_ylabel('summed spectral error'); ax.set_title('Error per frequency band')
ax.grid(True, which='both', axis='y', alpha=0.3); ax.legend()

fig.suptitle(f'Spectral error analysis — KAN + refinement (t = {t_spec:.3f})', fontsize=14)
plt.tight_layout()
fig.savefig(os.path.join(FIGDIR, 'burgers_kan_refine_methods_spectral.pdf'), bbox_inches='tight')
print('saved ->', os.path.join(FIGDIR, 'burgers_kan_refine_methods_spectral.pdf'))
plt.show()

band_df = pd.DataFrame(band_err, index=band_labels)
print('\n=== summed |spectral error| per band (final time) ===')
print(band_df.to_string())
```

## 9. Notes

- **Adaptive grid refinement** is configured entirely through the architecture params
  (`PARAMS_TD` / `PARAMS_VAN`): `refine_every`, `refine_factor` (default +10 %), `refine_tol`, and
  `refine_max_grid` (the cap on grid intervals). `refine_every=None` turns the feature off and recovers
  the plain KAN run — useful for an ablation (compare against
  `burgers_tdpinns_vs_vanilla_sawtooth_kan.ipynb`).
- **When it fires**: only when the loss is *not* reducing fast enough — window improvement below
  `refine_tol` over `refine_every` epochs. The training logs print each refinement
  (`grid refined -> N`), and `refine_report` / the final-grid printouts summarise how far each net grew.
- The trainer rebuilds the optimiser after each refinement (the spline coefficients are new tensors);
  the lr restarts from `max_lr`, which helps the optimiser adapt the freshly added knots.
- Everything else matches the KAN notebook: residuals, weighting, collocation, `TimeSlice`, and
  `compute_metrics`. Figures saved to `figs/burgers_kan_refine_methods_*.pdf`.
- `pykan` writes a `./model` dir on every construction **and every refinement** — `clear_kan_cache()`
  keeps it from accumulating between runs.
