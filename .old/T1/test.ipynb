```python

import json


class TranslationTool:

    def __init__(self):
        self.models = ["o3","o4","gpt5"]
        pass

    
    # BEGIN - INCLUIR NUEVO

    def eval_words(text: str) -> bool:
        for ch in text:
            if ch.isalpha():          # Unicode-aware test: works for all scripts
                return True
        return False
    
    # END - INCLUIR NUEVO

    def generate_translation_suggestions(self, text_array, output_language, image=None, temperature=0.3, top_p=1.0, request_data=None):
        # prompt = self.get_prompt(request_data, image)

        system_prompt = " " #prompt.format(output_language=output_language)

        array_input = [{"id": id, "original_text": text} for id, text in enumerate(text_array)]


        # BEGIN - INCLUIR NUEVO

        mask_array = [ TranslationTool.eval_words(item["original_text"]) for item in array_input  ] # máscara con palabras que deben enviarse a la LLM

        array_input_masked = [ array_input[i] for i in range(len(array_input)) if mask_array[i] ] # lista descartando palabras que no se traducen

        # END - INCLUIR NUEVO

        # BEGIN - MODIFICACION

        json_text = json.dumps(array_input_masked, indent=4) # ahora no se toma array_input sino array_input_maked

        # END - MODIFICACION

        suggestions = [[] for _ in range(len(text_array))]
        logs = []

        for model in self.models:
            translations = self.generate_chat_completion(
                system_prompt, array_input_masked, image, model=model, temperature=temperature, top_p=top_p
            )
            # BEGIN - MODIFICACION

            i_masked = 0 # contador para las palabras que si fueron traducidas

            for i in range(len(mask_array)):
                if mask_array[i]: # si en la mascara la palabra era valida para traduccion, se agrega directamente
                    suggestions[i].append(translations["json_response"][i_masked]["translated_text"])
                    i_masked += 1
                else: # sino se agrega la palabra original
                    suggestions[i].append(array_input[i]["original_text"])

            # END - MODIFICACION
            
                # create_log_data = self.create_log_data(request_data, translations, model, i)
                # logs.append(create_log_data)

            
        return dict(suggestions=suggestions, logs=logs)


    def generate_chat_completion(self,p1_, array_input_masked, p3_, model="", temperature="", top_p=0):

        response = []

        for item in array_input_masked:
            item_edited = item
            item_edited["translated_text"] = "TRANSLATED_"+item["original_text"]
            response.append(item_edited)

        return {"json_response":response}



```
