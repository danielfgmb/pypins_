# Burgers 2-D

```python
import TD_PINNS

TD_PINNS.set_environment(space_dimensions = 2, output_dimensions = 2, constants = 3) # Numero de inputs vs outputs

u_t = TD_PINNS.time_derivative(TD_PINNS.output[0]) # $$ \frac{\partial u}{\partial t} $$

u_x = TD_PINNS.derivative(TD_PINNS.output[0],TD_PINNS.x[0]) # $$ \frac{\partial u}{\partial x} $$

u_xx = TD_PINNS.derivative(u_x, TD_PINNS.x[0]) # $$ \frac{\partial^2 u}{\partial x^2} $$

u_y = TD_PINNS.derivative(TD_PINNS.output[0],TD_PINNS.x[1]) # $$ \frac{\partial u}{\partial y} $$

u_yy = TD_PINNS.derivative(u_y, TD_PINNS.x[1]) # $$ \frac{\partial^2 u}{\partial y^2} $$

v_t = TD_PINNS.time_derivate(TD_PINNS.output[1]) # $$ \frac{\partial v}{\partial t} $$

v_x = TD_PINNS.derivative(TD_PINNS.output[1],TD_PINNS.x[0]) # $$ \frac{\partial u}{\partial x} $$

v_xx = TD_PINNS.derivative(v_x,TD_PINNS.x[0]) # $$ \frac{\partial^2 v}{\partial x^2} $$

v_y = TD_PINNS.derivative(TD_PINNS.output[1],TD_PINNS.x[1]) # $$ \frac{\partial u}{\partial x} $$

v_yy = TD_PINNS.derivative(v_y,TD_PINNS.x[1]) # $$ \frac{\partial^2 v}{\partial y^2} $$

```

```python


eqn1 = u_t + TD_PINNS.output[0]*u_x + TD_PINNS.output[1]*u_y - TD_PINNS.constant[0]*(u_xx + u_yy) # paso todos los terminos a restar

eqn2 = v_t + TD_PINNS.output[0]*v_x + TD_PINNS.output[1]*v_y - TD_PINNS.constant[0]*(v_xx + v_yy) # paso todos los terminos a restar

TD_PINNS.set_governing_equations([eqn1,eqn2])
```

```python
TD_PINNS.train()

TD_PINNS.evaluate(t=0, x=[0,0.3]) # => []
```
