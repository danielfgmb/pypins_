```python
import matplotlib.pyplot as plt

# Datos extraídos del experimento
x = [0, 20, 40, 60, 80]  # tamaño de input
y = [0.789, 17.050, 38.386, 58.956, 78.335]  # tiempo en segundos

plt.figure(figsize=(8,5))
plt.plot(x, y, marker='o', linestyle='-', linewidth=2)

plt.title("Tiempo de ejecución vs Tamaño de input (Claude Sonnet 4)")
plt.xlabel("Tamaño del input (X words)")
plt.ylabel("Tiempo de ejecución (segundos)")
plt.grid(True)
plt.show()
```
